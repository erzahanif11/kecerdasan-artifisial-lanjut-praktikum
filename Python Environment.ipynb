```python
harga_dasar = 50000
harga = harga_dasar
umur = int(input("Umur: "))
hari = str(input("Hari beli: ").lower())
if (umur < 12 or umur > 60):
  harga *= 50/100
if (hari == "sabtu" or hari == "minggu"):
  harga += 10000
elif (umur >= 12 and umur <= 60) and (hari in ["senin", "selasa", "rabu", "kamis", "jumat"]):
  harga -= 5000
print(harga)
```

### Pertanyaan 3

```python
n = 4
for i in range(1, n + 1):
  for j in range(n - i):
    print(" ", end="")
  for k in range(i):
    print("* ", end="")
  print()
```

#### **Penjelasan**

Menggunakan loop bersarang dengan perulangan i sebagai loop utama untuk baris yang melakukan looping dimulai dari 1 hingga n + 1 atau 5. Perulangan kedua yakni untuk mencetak spasi sebanyak n - i pada tiap baris. Perulangan ketiga mencetak bintang sebanyak i.
